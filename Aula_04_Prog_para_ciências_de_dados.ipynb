```python
#for para passar pela lista
# Lista contendo as notas das provas
notas_provas = [9,6.5,3.4,8.5]
# Variável acumuladora, começa em 0 para poder somar as notas depois
soma = 0
# Laço for percorre cada elemento da lista diretamente (sem precisar de índice)
# a cada volta, 'nota' recebe um valor da lista (9, depois 6.5, depois 3.4, depois 8.5)
for nota in notas_provas :
   #a cada iteração, soma o valor atual de 'nota' ao total acumulado em 'soma'
  soma = soma + nota
# fora do for (sem indentação) -> só executa depois que o laço terminar
# calcula a média dividindo a soma total pela quantidade de notas
média = soma / 4
# imprime o texto e o valor da média calculada
print ("Media = ", média)
```

```python
#usamos
def calcula_percento (valor, percentual = 100):
  quanto = valor * percentual / 100
 # devolve o resultado para quem chamou a função
    # e encerra a execução da função nesse ponto
  return quanto
# chama a função passando valor=300 e percentual=20
# dentro da função: quanto = 300 * 20 / 100 = 60.0 -> return 60.0
# .format() insere esse valor no {:10.2f} (float, 2 casas decimais, largura 10)
print ("20% de 300 são: {:10.2f}" .format (calcula_percento (300,20)))
print ("100% de 50 são : {:10.2f}" .format(calcula_percento (50)))
```

```python
# Importa o módulo math, que traz funções e constantes matemáticas prontas
# math.pi é uma constante já pronta com o valor de pi (3.14159265...)
import math
# f antes das aspas cria uma f-string (string formatada)
# tudo que está dentro de {} é substituído pelo valor real da variável/expressão
# math.pi -> valor que será inserido no texto
# :.4f -> formatação: 'f' = número float (decimal), '.4' = 4 casas decimais
# ou seja: pega o valor de pi e arredonda para 4 casas decimais (3.1416)
print(f"o valor de pi é: {math.pi: .4f}.")
```

```python
# Importa math para poder usar a função sqrt() pronta no final, para comparação
import math

# Lê o número que queremos calcular a raiz quadrada
n = float(input('Digite um número para calcular a raiz quadrada: '))

# 'b' é o chute inicial (aproximação inicial de raiz quadrada)
# pode-se começar com qualquer valor > 0; aqui usamos o próprio n como início
b = n

# calcula o primeiro valor de p usando a fórmula de Newton: p = (b + n/b) / 2
p = (b + (n / b)) / 2

# repete enquanto a diferença absoluta entre n e p² for maior ou igual a 0.0001
# abs() garante que pegamos a diferença sempre positiva (não importa o sinal)
while abs(n - (p ** 2)) >= 0.0001:
    # antes de recalcular p, guardamos o valor atual em b
    # (b vira o 'p anterior', para ser usado na próxima fórmula)
    b = p

    # recalcula p usando a fórmula de Newton, agora com o novo b
    p = (b + (n / b)) / 2

# quando o while para, significa que p já está bem próximo da raiz real de n

# calcula a raiz quadrada "oficial" usando a função pronta do Python
raiz_math = math.sqrt(n)

# mostra os dois resultados para comparação
print(f'Raiz quadrada calculada pelo método de Newton: {p:.4f}')
print(f'Raiz quadrada calculada pela função sqrt(): {raiz_math:.4f}')
```
