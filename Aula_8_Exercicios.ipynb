```python
import pandas as pd
```

```python
#Importando um arquivo .csv dos dados de temperatura dos municípios
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
```

```python
1. Localizar a cidade e mostrar todas as informações
```

```python
santo_cristo = df[df['nome_municipio'] == 'Santo Cristo']
santo_cristo
```

```python
2. Qual o mês de temperatura mínima
```

```python
# colunas de temperatura mínima
colunas_tmin = ['tmin_jan','tmin_fev','tmin_mar','tmin_abr','tmin_mai',
                'tmin_jun','tmin_jul','tmin_ago','tmin_set','tmin_out','tmin_nov','tmin_dez']

# pega os valores de Santo Cristo
valores_tmin = santo_cristo[colunas_tmin].values[0]

# encontra o mês com menor valor
mes_tmin = colunas_tmin[valores_tmin.argmin()]
print(f'Mês de temperatura mínima: {mes_tmin} ({valores_tmin.min():.2f}°C)')
```

```python
3. Qual o mês de maior déficit hídrico
```

```python
# colunas de déficit hídrico
colunas_def = ['def_jan','def_fev','def_mar','def_abr','def_mai',
               'def_jun','def_jul','def_ago','def_set','def_out','def_nov','def_dez']

valores_def = santo_cristo[colunas_def].values[0]

mes_def = colunas_def[valores_def.argmax()]
print(f'Mês de maior déficit hídrico: {mes_def} ({valores_def.max():.2f} mm/mês)')
```

```python
4. Média de temperatura pelas 4 estações
```

```python
# Hemisfério Sul:
# Verão: dez, jan, fev
# Outono: mar, abr, mai
# Inverno: jun, jul, ago
# Primavera: set, out, nov

verao    = santo_cristo[['temp_media_dez','temp_media_jan','temp_media_fev']].mean(axis=1).values[0]
outono   = santo_cristo[['temp_media_mar','temp_media_abr','temp_media_mai']].mean(axis=1).values[0]
inverno  = santo_cristo[['temp_media_jun','temp_media_jul','temp_media_ago']].mean(axis=1).values[0]
primavera= santo_cristo[['temp_media_set','temp_media_out','temp_media_nov']].mean(axis=1).values[0]

print(f'Verão:     {verao:.2f}°C')
print(f'Outono:    {outono:.2f}°C')
print(f'Inverno:   {inverno:.2f}°C')
print(f'Primavera: {primavera:.2f}°C')
```
